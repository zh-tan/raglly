```python
import openai


# Builds the prompt for a specific speaker
def build_prompt_speaker(context, grc, party, members, speaker):
    prompt = f"""Context: This is a Singapore Election 2025 Rally at {grc} GRC for the party {party}. The members running are: {", ".join(members)}.
Below is the transcript from a YouTube video. Some parts may be inaccurate due to transcription errors.
Summarize what this speaker, {speaker}, has said.

Transcript:
{context}"""
    
    return [{"role": "user", "content": prompt}]

# Main summarization loop
def summarize_speakers():
    summaries = {
        'pap': {},
        'wp': {}
    }

    for party, grc_data in speakers_compile.items():
        for grc, speakers in grc_data.items():
            member_names = [s["name"] for s in speakers if "Candidate" in s["role"]]
            for speaker in speakers:
                messages = build_prompt_speaker(
                    context=transcript_mapping[party][grc],
                    grc=grc,
                    party=party.upper(),
                    members=member_names,
                    speaker=speaker["name"]
                )

                print(f"Summarizing {speaker['name']}...")

                response = client.chat.completions.create(
                    model="gpt-4.1",
                    messages=messages,
                    temperature=0.2
                )
                
                
                summaries[party][speaker["name"]] = response.choices[0].message.content

    return summaries

# Example usage
# full_transcript = open("transcript.txt").read()
results = summarize_speakers()
# for speaker, summary in results.items():
#     print(f"\n--- {speaker} ---\n{summary}\n")
```

```python
for party, speakers in results.items():
    for speaker, summary in speakers.items():
        # Create a safe filename by replacing spaces with underscores
        filename = f"{speaker.replace(' ', '_')}_{party}_punggol_summary.txt"
        with open(filename, "w", encoding="utf-8") as f:
            f.write(summary)

        print(f"Saved: {filename}")
```

```python
def build_llm_prompt(user_query, search_results):
        
    payload_q = {
    "query": user_query,
    # "filters": {
    #     "type": "eq",
    #     "key": "party",
    #     "value": "WP (Worker's Party)"
    # }
    }
    
    response = requests.post(url, headers=headers, json=payload_q)

    context_blocks = []

    for result in search_results["data"]:
        score = result['score']
        
        # threshold
        if score < 0.75:
            continue 
            
        speaker = result["attributes"]["speaker"]
        party = result["attributes"]["party"]
        grc = result["attributes"]["grc"]
        content = result["content"][0]["text"]

        context_blocks.append(f"""
### Speaker: {speaker}
**Party:** {party}  
**GRC:** {grc}  
**Summary:**  
{content.strip()}
""")

    context_text = "\n".join(context_blocks)

    prompt = f"""You are an assistant analyzing speeches from the Singapore General Election 2025 rallies.

The user has asked: **{user_query}**

Below are summaries of what different speakers from different parties have said:

{context_text}

---  
Based on the above, please answer the user's prompt and summarize the key messages or contrasts across all the speakers. Present your response clearly and concisely, ideally structured by theme or by party.
"""

    return [{"role": "user", "content": prompt}]
```
